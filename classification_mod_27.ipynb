# Verification of mod 27 classification of eigenforms
For the faster option of using the precomputed source data, set USE_ARCHIVED_SOURCE_DATA = True, Otherwise, set it to FALSE to recompute the source data.

```python
import sys
from pathlib import Path

PYTHON_DIRECTORY = Path("python").resolve()
if str(PYTHON_DIRECTORY) not in sys.path:
    sys.path.insert(0, str(PYTHON_DIRECTORY))

from hecke_congruences import *
from load_source_data import load_source_data

USE_ARCHIVED_SOURCE_DATA = True
SOURCE_DATA_DIRECTORY = Path("source_data")
MOD27_SOURCE_ARCHIVE = (
    SOURCE_DATA_DIRECTORY / "p3_mod27_T2_T7_all_degrees.npz"
)
MOD27_ZERO_SOURCE_ARCHIVE = (
    SOURCE_DATA_DIRECTORY / "p3_mod27_zero_T2_mod243.npz"
)

def get_source_data(R, d, q, archive):
    if USE_ARCHIVED_SOURCE_DATA:
        return load_source_data(R, d, q, archive)

    return prepare_source_data(R, d, q)
```

# Identities for $T_7$

```python
p = 3
m = 3
modulus = p^m
R = Integers(modulus)

S.<X> = PolynomialRing(R)

period = euler_phi(p^m)       
a_m = p^m * (p - 1)    
b_m = p^(m - 1)*(p + 1) 
surjectivity_bound = a_m + b_m  


base_degrees = tuple( range(0, a_m + b_m, 2))
degree_residues = tuple(d % period for d in base_degrees)

F7 = {r : X - (1 + 7^(r+1)) for r in degree_residues}

exact_induction_base = tuple(
    range(
        p**(m - 1)*(p + 1),
        a_m + b_m,
        2,
    )
)

for d in base_degrees:
    for q in range(0, p-1):
        data = get_source_data(R, d, q, MOD27_SOURCE_ARCHIVE)

        test = verify_ordinary_identities(
            F=F7[d%period],
            n=7,
            data=data,
        )

        print(test)
        assert test["passed"]

print("——————————————————————————————————————————————")
print("IDENTITY VERIFIED")
```

# Identities for $T_2$ - non-zero branch

```python
p = 3
m = 3
modulus = p^m
R = Integers(modulus)

S.<X> = PolynomialRing(R)

period = euler_phi(p^m)       
a_m = p^m * (p - 1)    
b_m = p^(m - 1)*(p + 1) 
surjectivity_bound = a_m + b_m  


base_degrees = tuple( d for d in  range(0, a_m + b_m, 2) if d%6 != 2)
degree_residues = tuple(d%period for d in base_degrees)

cr = {
    0: 1,
    4: 4,
    6: 4,
    10: 1,
    12: 7,
    16: 7,
}

F2 = {r : X^2 - cr[r] for r in degree_residues}

exact_induction_base = tuple(
    range(
        p**(m - 1)*(p + 1),
        a_m + b_m,
        2,
    )
)

print("Dickson degrees:", a_m, b_m)
print("full finite verification range:", base_degrees)
print("exact induction base:", exact_induction_base)

for d in base_degrees:
    for q in range(0,p-1):
        data = get_source_data(R, d, q, MOD27_SOURCE_ARCHIVE)

        test = verify_divided_identities(
            F=F2[d%period],
            Q=X,
            n=2,
            a=1,
            b=2,
            data=data,
        )

        print(test)
        assert test["passed"]

print("——————————————————————————————————————————————")
print("IDENTITY VERIFIED")
```

# Identities for $T_2$ - zero branch

```python
p = 3
m = 5
modulus = p^m
R = Integers(modulus)

S.<X> = PolynomialRing(R)

period = euler_phi(p^m)
a_m = p^m * (p - 1)
b_m = p^(m - 1)*(p + 1)
surjectivity_bound = a_m + b_m


base_degrees = tuple(d for d in range(0, a_m + b_m, 2) if d%6 == 2)

degree_residues = tuple(d%period for d in base_degrees)

F2 = {
    r: (X*(X + 1))^2
    for r in degree_residues
}

Q2 = X*(X - 63)

exact_induction_base = tuple(
    d
    for d in range(
        p**(m - 1)*(p + 1),
        a_m + b_m,
        2,
    )
    if d%6 == 2
)

print("Dickson degrees:", a_m, b_m)
print("full finite verification range:", base_degrees)
print("exact induction base:", exact_induction_base)
```

```python
def verify_zero_branch_case(d, q):
    """
    Verify the zero-branch staged T2 identity in one degree and
    orientation.

    Source data are loaded from the archive or computed afresh,
    according to USE_ARCHIVED_SOURCE_DATA.
    """
    data = get_source_data(
        R,
        d,
        q,
        MOD27_ZERO_SOURCE_ARCHIVE,
    )

    return verify_divided_identities(
        F=F2[d % period],
        Q=Q2,
        n=2,
        a=4,
        b=1,
        data=data,
        check_descent=False,
    )


lower_base_degrees = tuple(
    d
    for d in base_degrees
    if d not in exact_induction_base
)

ordered_degrees = (
    tuple(sorted(exact_induction_base, reverse=True))
    + tuple(sorted(lower_base_degrees, reverse=True))
)

cases = [
    (d, q)
    for d in ordered_degrees
    for q in range(0, p - 1)
]
results = []

for d, q in cases:
    test = verify_zero_branch_case(d, q)
    results.append(test)

    print(
        f"d={d:3d}, "
        f"r={test['residue']:3d}, "
        f"q={q}, "
        f"sign={test['sign']:+d}, "
        f"rank={test['rank']:3d}, "
        f"division={test['division_passed']}, "
        f"terminal={test['terminal_passed']}, "
        f"passed={test['passed']}"
    )

    assert test["passed"]

print("——————————————————————————————————————————————")
print("cases completed:", len(results))
print("ALL ZERO-BRANCH IDENTITIES VERIFIED")
```

# Check that each possible signature given by the identities above is realized by a strong eigenform

```python
def krw_reduction_to_integer(x, nf, pr, p, m):
    e = pr[2]
    N = e*(m - 1) + 1

    candidates = [
        r
        for r in range(p^m)
        if nf.idealval(x - r, pr) >= N
    ]

    if len(candidates) != 1:
        raise ValueError(
            f"expected one residue in Z/{p^m}Z, "
            f"but found {candidates}"
        )

    return Integers(p^m)(candidates[0])
```

```python
B = 70
p = 3
m = 3
l1, l2 = 2, 7

R = Integers(p^m)
signatures = []

for k in range(12, B + 1, 2):
    S = CuspForms(1, k)

    if S.dimension() == 0:
        continue

    print(f"calculating strong signatures at weight {k}")

    for j, f in enumerate(S.newforms(names='a')):
        K = f.base_ring()

        # Rational eigenform orbit
        if K == QQ:
            signature = (
                k,
                R(f[l1]),
                R(f[l2]),
            )
            signatures.append(signature)
            print(signature)
            continue

        # Nonrational eigenform orbit
        pol = K.pari_polynomial('y')
        nf = pari([pol, [p]]).nfinit(4)

        a1 = pari(f[l1])
        a2 = pari(f[l2])

        for pr in nf.idealprimedec(p):
            e = pr[2]
            N = e*(m - 1) + 1
            I = nf.idealpow(pr, N)

            a1_bar = krw_reduction_to_integer(
                a1, nf, pr, p, m
            )

            a2_bar = krw_reduction_to_integer(
                a2, nf, pr, p, m
            )

            signature = (k, a1_bar, a2_bar)
            signatures.append(signature)
```

```python
signatures_by_weight_residue = {}
period = euler_phi(p^m)
for s in signatures:
    r = s[0]%period
    if not r in signatures_by_weight_residue:
        signatures_by_weight_residue[r] = set()
    signatures_by_weight_residue[r].add((s[1],s[2]))
```

```python
print(
    f"{'k mod ' + str(period):<10} | "
    f"(a_{l1}, a_{l2}) mod {p^m}"
)
print("-" * 70)

for r in sorted(signatures_by_weight_residue):
    pairs = ", ".join(
        f"({a1}, {a2})"
        for a1, a2
        in sorted(signatures_by_weight_residue[r])
    )

    print(f"{str(r):<10} | {pairs}")
```

## Compare with the cyclotomic packets

```python
cyclotomic_packets = {
    0:  [(0, 17), (8, 9)],
    2:  [(0, 1),  (9, 10)],
    4:  [(0, 3),  (6, 15), (9, 12)],
    6:  [(0, 5),  (9, 14)],
    8:  [(0, 7),  (9, 16)],
    10: [(0, 9),  (3, 6),  (12, 15)],
    12: [(0, 11), (2, 9)],
    14: [(0, 13), (4, 9)],
    16: [(0, 15), (3, 12), (6, 9)],
}

cyclotomic_signatures = {}

for r, packets in cyclotomic_packets.items():
    cyclotomic_signatures[r] = {
        (
            ZZ(R(l1)^a + R(l1)^b),
            ZZ(R(l2)^a + R(l2)^b),
        )
        for a, b in packets
    }

print(
    f"{'k mod ' + str(period):<10} | "
    f"(a_{l1}, a_{l2}) mod {p^m}"
)
print("-" * 70)

for r in sorted(cyclotomic_signatures):
    pairs = ", ".join(
        f"({a1}, {a2})"
        for a1, a2 in sorted(cyclotomic_signatures[r])
    )

    print(f"{str(r):<10} | {pairs}")
```

```python
assert cyclotomic_signatures == signatures_by_weight_residue
print("EVERY POSSIBLE MODULO-27 SIGNATURE IS REALIZED")
```
